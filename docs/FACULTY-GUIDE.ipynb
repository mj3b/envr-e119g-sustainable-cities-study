Record what decision each document makes and the question it addresses. Then consider this proposed sentence: “Because the request was late, the water concern never received an answer.” Does the later decision support that sentence? What would be required to assess whether the answer was adequate?

## Assess the practical proposal

The proposed register would link a concern to an official response, the responsible body and documented follow-through. It would preserve original records and identify missing evidence. Its status is a proposal for testing.

| Review question | Evidence that would inform it |
| --- | --- |
| Can residents find and understand a response? | A defined task trial, recorded mistakes and assistance, and recruitment limits |
| Can staff keep the index current? | Update effort, source-link failures and correction handling |
| Were affected interests represented? | Evidence about participation and those absent from the record |
| Did water conditions improve? | Comparable local measurements and an assessment of alternative explanations |

The first two questions describe the proposed pilot. The latter two require a broader research design. A readable register does not establish representative participation or environmental improvement.

## Separate completed checks from scholarly judgments

The research record reports the November 14 software verification and its limits. The fixed-rule graph reproduced stored annotations and kept candidate results separate from the human-review record. Fifteen author decisions followed AI suggestions and translations. No independent or blinded reviewer is recorded.

A source-file hash identifies bytes. A software test checks behavior for its inputs. A rendering check can detect clipping or broken tables. None of those checks establishes that a sentence faithfully represents a source or that a reader understands the argument. The v0.4 revision explicitly responds to the author’s difficulty following the preceding memo.

The [research-integrity protocol](../methods/RESEARCH-INTEGRITY.md) and [AI-system register](../methods/AI-SYSTEMS.md) define the roles. Human fidelity is not inferred from publication, conversion to notebooks or a successful software check.

## Record feedback that can change the study

Choose one bounded point for comment. Quote the memo sentence, name its source and page, and explain whether the issue concerns evidence, inference or presentation. For the proposal, identify the institution, resource or information requirement that needs clarification.

Useful questions include whether Sen’s concept explains a specific part of the inquiry, whether the register has a plausible custodian, and which missing record would change a conclusion about resident influence. The [writing rules](../standards/WRITING.md) ask for visible reasoning and qualifications rather than claims of quality based on polish.
