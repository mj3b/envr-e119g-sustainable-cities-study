## Try one source-reading exercise

Open the [tribunal judgment at PDF page 64](https://tribunalambiental.cl/wp-content/uploads/2024/02/2024.02.26_Sentencia_R-271_270-2020.pdf#page=64). Consideration 93 discusses a proposed change to cooling equipment. The judgment is in Spanish; the memo’s research record supplies an identified excerpt and an AI-assisted English translation.

Write down the actor, the proposed action and the date described. Then decide which question this passage can answer.

1. Was a cooling change proposed?
2. Was the equipment installed?
3. Did nearby users experience measured water savings?

Use the PDF viewer’s page count. Court folio numbers and PDF page numbers can differ. Keep the Spanish locator attached to any translation or paraphrase.

## Compare your reasoning

The passage records the proposed change. Installation would require evidence of what was built. Measured savings would require observations, a stated comparison and evidence of who benefited. A proposal, an installed system and an observed benefit are different claims.

| Evidence you might find | What it could establish | What needs a further check |
| --- | --- | --- |
| Dated design proposal | What the developer proposed | Whether it was authorized and built |
| Installation or inspection record | What was installed within its stated scope | Water withdrawn during operation |
| Meter record | Recorded withdrawals for a stated period | Comparable baseline and effects on other water users |

This table is a reading aid, not a list of records already obtained for Cerrillos. The study’s [current question and limits](../research/CURRENT.md) identify the bounded case conclusion.

## Apply the same care to a resident’s concern

A ruling on whether a request can proceed answers a procedural question. A response discussing groundwater answers a question about official treatment of the concern. Evidence of an environmental outcome answers a further question. Assignment 4 follows those stages separately.

Try the memo’s proposed register entry. Can you locate the concern, the authority’s answer and what remains unknown? If a field is missing, write down the document or responsible office that could supply it. A clear index could help a reader find an answer. Its usefulness and maintenance burden would still need testing.

The selected filings cannot establish every resident’s view or whether the recorded concerns caused a design decision. Those are questions for additional evidence.
