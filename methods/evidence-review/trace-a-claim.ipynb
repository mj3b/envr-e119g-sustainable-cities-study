# Trace a water claim to its source

A city announces how much water its cooling project expects to use in July. What evidence would let you write that the city **measured** its July withdrawals?

This lesson uses invented records. You can read the passages and saved output on GitHub. For the wider study, return to the [reader guide](../../docs/START-HERE.md).


## Try a label before checking the output

The sentence to check is “The city measured the project's July water withdrawals.” Treat the first three rows as separate scenarios.

| Case | What the record says | What can you infer? |
| --- | --- | --- |
| T01 | “The city took no meter readings of project water withdrawals in July.” | Does the record directly answer the claim? |
| T02 | “The project plans to withdraw 100 cubic metres of water in July.” | What kind of number is 100? |
| T03 | “The city measured 92 cubic metres of project water withdrawals in July.” | What action and month does this record document? |
| T04 | The July meter report is the only evidence for a claim about both July and August. | Can the report support each month? |

Choose **Supported**, **Contradicted** or **Insufficient**. For the two-month claim, check each month first.


## See how the workflow handles the evidence

The [case file](cases.json) records an analyst’s proposed relation between each clause and its passage. The [specification](SPEC.md) explains how those relations produce a candidate label. LangChain Core keeps each excerpt with its source details. LangGraph applies the steps in order and routes every result to a person.

To run the example locally, follow the [setup instructions](README.md#run-the-demonstration-checks). Open this notebook in Jupyter from the example folder and run the cell below. You can change `claim_id` to T01, T02 or T03 to see its evidence.


```python
from pathlib import Path

from workflow import run

packet = run(Path("cases.json").read_bytes())
for claim in packet["claims"]:
    print(f'{claim["id"]}: {claim["candidate_label"]} | {claim["review_status"]}')

claim_id = "T04"
claim = next(row for row in packet["claims"] if row["id"] == claim_id)
print(f'\n{claim["text"]}')
for clause in claim["clauses"]:
    print(f'{clause["id"]}: {clause["candidate_label"]}')
    for link in clause["evidence"]:
        print(f'  {link["passage_id"]} / {link["source"]["id"]} | {link["locator"]}')
        print(f'  {link["rationale"]}')
```

## Compare your reasoning

The city’s explicit statement in T01 directly contradicts the measurement claim. T02 leaves it insufficient because the design plan projects use. T03 supports the statement for July. In T04, the July clause is supported and the August clause is insufficient. The combined claim is insufficient.
